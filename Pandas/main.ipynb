```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv("vendas_tech.csv", encoding="utf-8", low_memory=False, sep=",")
#display(df_vendas)
display(df_vendas.info())

df_analise = df_vendas.drop(columns="Data_Base")
df_analise["Loja"] = df_analise["Loja"].fillna("Online")
df_analise["Loja"] = df_analise["Loja"].str.strip()
df_analise["Loja"] = df_analise["Loja"].str.title()
df_analise["Data"] = pd.to_datetime(df_analise["Data"], format="%Y-%m-%d")
df_analise = df_analise.drop_duplicates(subset="ID_Pedido", keep="last")

display(df_analise.info())
display(df_analise)

display(df_analise["Preco_Unitario"].describe())
```

```python
import pandas as pd
import numpy as np

df_data = pd.read_csv('data.csv')
display(df_data.info())
df_analises = df_data

df_analises["cidade"] = df_analises['cidade'].fillna("Não Informado")
df_analises["cidade"] = df_analises['cidade'].str.strip().str.title()
df_analises["data_compra"] = pd.to_datetime(df_analises["data_compra"], errors="coerce", format="%Y-%m-%d")
df_analises["data_compra"] = df_analises["data_compra"].fillna("Não Informado")
df_analises["idade"] = pd.to_numeric(df_analises["idade"], errors="coerce")
df_analises["idade"] = df_analises["idade"].fillna(df_analises["idade"].mean())
df_analises["idade"] = df_analises["idade"].astype(int)
df_analises["valor_compra"] = pd. to_numeric(df_analises["valor_compra"], errors="coerce")
df_analises["produto"] = df_analises["produto"].fillna("Não Informado")
df_analises = df_analises.dropna(subset='valor_compra')


display(df_analises)
display(df_analises.info())

```
